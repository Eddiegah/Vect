Scientific computing as native language syntax.

```vect
var v1 = [1.0, 2.0, 3.0]
var v2 = [4.0, 5.0, 6.0]
print(v1 + v2)
print(v1 · v2)
```

```vect
var A = [[1.0, 2.0], [3.0, 4.0]]
var B = [[5.0, 6.0], [7.0, 8.0]]
print(A @ B)
print(det(A))
```

```vect
sym f(x) = x**2 + 3*x + 1
var df = d/dx(f(x))
print(df)
print(eval(df, x=2.0))
```

```vect
sym height(t) = 20.0*t - 4.9*t**2
var velocity = d/dt(height(t))
print(velocity)
var area = integral(height(t), "t", 0.0, 2.0)
print(f"Distance in 2s: {area} metres")
```

```vect
import "stdlib/physics.vect"
print(kinetic_energy(70.0, 10.0))
print(f"Momentum: {momentum(70.0, 10.0)} kg*m/s")
```

```vect
sym wave(x) = sin(x) * 2.0 + 0.5*sin(3.0*x)
plot(wave(x), x, -6.28, 6.28, "Wave Function")
print("Plot saved to vect_plot.png")
```
